```python
employee={"101":{
    'name':'zaheer',
    'age':19,
    'Department':'IT',
    'salary':50000}}

```

```python
def add_employee(id_no,name,age,department,salary):
    if id_no in employee:
        print(f"employee with ID({id_no}) already exist")
    else:
         print("new employee added/n")
         employee[id_no]={
         'name':name,
         'age':age,
        'Department':department,
         'salary':salary
          }

    print( id_no,  employee[id_no])
```

```python
def view_all_employees():
    if not employee:
        print("No employees yet.\n")
        return

    for id_no, details in employee.items():            # <- .items() (plural)
        print("ID:",id_no)
        for key, value in details.items():              # <- .items() (plural)
            print(f"  {key}: {value}")
        print("-----------------------------")
```

```python
def search_employee(key):
    if key in employee :
     print( key, employee[key])
    else :
        print("Employee doesn't exist")
```

```python
# ---------------------------
# 1. DATA TYPES & VARIABLES
# ---------------------------
# Numeric data types
integer_num = 10 # int
float_num = 3.14 # float
complex_num = 2 + 3j # complex
# String data type
name = "Sara"
# Boolean data type
is_student = True
# Collection data types
fruits = ["apple", "banana", "mango"] # list
coordinates = (10, 20) # tuple
unique_nums = {1, 2, 3, 3} # set
student = {"name": "Sara", "age": 20} # dictionary
print("Integer:", integer_num, "Type:", type(integer_num))
print("Float:", float_num, "Type:", type(float_num))
print("Complex:", complex_num, "Type:", type(complex_num))
print("Name:", name, "Type:", type(name))
print("Boolean:", is_student, "Type:", type(is_student))
print("List:", fruits)
print("Tuple:", coordinates)
print("Set:", unique_nums)
print("Dictionary:", student)
print("--------------------------------------------------")
# ---------------------------
# 2. OPERATORS
# ---------------------------
a = 15
b = 4
# Arithmetic operators
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
# Relational operators
print("a > b:", a > b)
print("a == b:", a == b)
# Logical operators
x = True
y = False
print("x and y:", x and y)
print("x or y:", x or y)
print("not x:", not x)
# Assignment operators
num = 10
num += 5 # same as num = num + 5
print("After += 5:", num)
# Membership operators
print("'apple' in fruits:", "apple" in fruits)
print("'grape' not in fruits:", "grape" not in fruits)
# Identity operators
list1 = [1, 2, 3]
list2 = [1, 2, 3]
print("list1 == list2:", list1 == list2) # compares values
print("list1 is list2:", list1 is list2) # compares memory location
print("--------------------------------------------------")
# ---------------------------
# 3. MINI APPLICATION
# ---------------------------
# Program to check if a student is eligible for voting
age = 18
citizen = True
if age >= 18 and citizen:
 print(name, "is eligible to vote.")
else:
 print(name, "is NOT eligible to vote.")
print("--------------------------------------------------")
```

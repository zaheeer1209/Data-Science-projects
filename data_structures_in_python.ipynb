```python
# Demonstration of Data Structures in Python
# 1. List
print("List Example:")
fruits = ["apple", "banana", "cherry"]
fruits.append("mango") # adding element
fruits.remove("banana") # removing element
print("Fruits List:", fruits)
print("Second fruit:", fruits[1]) # accessing element
print()
# 2. Tuple
print("Tuple Example:")
numbers = (10, 20, 30, 40)
print("Numbers Tuple:", numbers)
print("First element:", numbers[0])
# numbers[1] = 50 # ❌ This will give error since tuple is immutable
print()
# 3. Set
print("Set Example:")
colors = {"red", "green", "blue"}
colors.add("yellow") # add element
colors.discard("green") # remove element
print("Colors Set:", colors)
print("Is 'red' in set?", "red" in colors)
print()
# 4. Dictionary
print("Dictionary Example:")
student = {"name": "Sara", "age": 20, "course": "Python"}
print("Student Dictionary:", student)
print("Student Name:", student["name"])
student["age"] = 21 # update value
print("Updated Age:", student["age"])
print()
# 5. String
print("String Example:")
text = " Python Programming "
print("Original String:", text)
print("Stripped:", text.strip()) # remove spaces
print("Uppercase:", text.upper()) # convert case
print("Substring [0:6]:", text[0:6]) # slicing
print("Replaced:", text.replace("Python", "Java"))
```

```python
# Importing necessary libraries
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk import pos_tag

# Download required NLTK data files (run once)
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('punkt_tab') # Added to resolve LookupError

# Input text
text = """Natural Language Processing is a fascinating field of artificial intelligence. It enables machines to understand, process, and generate human language."""

# (A) Removing Stop Words
def remove_stopwords(tokens):
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
    return filtered_tokens

# (B) Stemming
def apply_stemming(tokens):
    stemmer = PorterStemmer()
    stemmed_tokens = [stemmer.stem(word) for word in tokens]
    return stemmed_tokens

# (C) POS Tagging
def pos_tagging(tokens):
    tagged = pos_tag(tokens)
    return tagged

# Main Function
def main():
    # Tokenization
    print("Original Text:\n", text, "\n")
    tokens = word_tokenize(text)
    print("Tokenized Words:")
    for token in tokens:
        print(token)
    print("\n")

    # (A) Remove Stop Words
    tokens_without_stopwords = remove_stopwords(tokens)
    print("Tokens after Stop Word Removal:")
    for token in tokens_without_stopwords:
        print(token)
    print("\n")

    # (B) Apply Stemming
    stemmed_tokens = apply_stemming(tokens_without_stopwords)
    print("Stemmed Tokens:")
    for token in stemmed_tokens:
        print(token)
    print("\n")

    # (C) POS Tagging
    pos_tags = pos_tagging(tokens_without_stopwords)
    print("POS Tags:")
    for word, tag in pos_tags:
        print(f"{word}: {tag}")

if __name__ == "__main__":
    main()
```

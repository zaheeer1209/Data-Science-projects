```python
# Concept: Analyze student marks using different loops
# For loop: input marks of 5 subjects
marks = []
print("Enter marks of 5 subjects:")
for i in range(1, 6):
 m = int(input(f"Subject {i}: "))
 marks.append(m)
# While loop: calculate total and average
i = 0
total = 0
while i < len(marks):
 total += marks[i]
 i += 1
 average = total / len(marks)
print("\nTotal Marks:", total)
print("Average Marks:", average)
# Nested loop: display marks as table (subject vs marks)
print("\nMarks Table:")
for i in range(1, len(marks)+1):
  for j in [marks[i-1]]:
    print(f"Subject {i} -> {j}")
print("------")
# Loop control: check for failed subjects
print("\nChecking results (skip 100, stop if any subject < 35):")
for mark in marks:
 if mark == 100:
  continue # skip checking if subject scored full marks
 if mark < 35:
  print("Fail detected! Stopping check.")
  break # stop loop if failed subject found
  print("Pass in subject with marks:", mark)
```
